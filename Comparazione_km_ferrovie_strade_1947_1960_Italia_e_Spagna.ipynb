# Spagna 1947-1960

```python
# Leggo dal mio drive la tabella spagnola sulle ferrovie
tab_spagna_ferrovie = pd.read_excel('Spagna_km_1947-1960.xlsx')
tab_spagna_ferrovie.head(10)
```

```python
# Leggo dal mio drive la tabella spagnola sulle ferrovie
tab_spagna_strade = pd.read_excel('Spagna_km_Strade_1947-1960.xlsx')
tab_spagna_strade.head(5)
```

```python
# Grafico a linee per comparazione con doppio asse Y
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot Km Strade on primary y-axis (ax1)
ax1.plot(tab_spagna_strade['Anni'], tab_spagna_strade['Km_Strade'], label='Km Strade Spagnole', color='blue')
ax1.set_xlabel('Anni')
ax1.set_ylabel('Lunghezza Strade (km)', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

# Create a secondary y-axis (ax2)
ax2 = ax1.twinx()
# Plot Km Ferrovie on secondary y-axis (ax2)
ax2.plot(tab_spagna_ferrovie['Anni'], tab_spagna_ferrovie['Totale_via_estrecha'], label='Km Ferrovie Secondarie Spagnole', color='red')
ax2.set_ylabel('Lunghezza Ferrovie (km)', color='red')
ax2.tick_params(axis='y', labelcolor='red')

# Add title and grid
plt.title('Confronto Km ferrovie secondarie e strade spagnole 1947-1960')
ax1.grid(True)

# Combine legends from both axes
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc='upper left')

plt.show()
```

# Ferrovie e basta 1945-1950

```python
import matplotlib.pyplot as plt

# Filter the Italian secondary railways data for the years 1947-1950
tab_italia_ferrovie_1947_1950 = tab_italia_ferrovie[(tab_italia_ferrovie['Anni '] >= 1947) & (tab_italia_ferrovie['Anni '] <= 1950)]

print("tab_italia_ferrovie_1947_1950 head:")
print(tab_italia_ferrovie_1947_1950.head())

# Create a line plot
plt.figure(figsize=(10, 6))
plt.plot(tab_italia_ferrovie_1947_1950['Anni '], tab_italia_ferrovie_1947_1950['Km_ferrovie'], marker='o', color='green')

# Add title and labels
plt.title('Km Ferrovie Secondarie Italiane (1947-1950)')
plt.xlabel('Anni')
plt.ylabel('Lunghezza Ferrovie (km)')
plt.grid(True)
plt.xticks(tab_italia_ferrovie_1947_1950['Anni ']) # Ensure all years are shown as ticks
plt.show()
```
